# Dessiner et reconnaître un chiffre

Ce notebook ouvre une petite fenêtre de dessin. Trace un seul chiffre avec la souris, clique sur **Prédire**, puis lis la classe estimée et la confiance du CNN.

Avant de l'utiliser, exécute le notebook d'entraînement jusqu'à sa cellule de sauvegarde. Sélectionne le kernel `.venv`.

```python
from pathlib import Path

import torch

from number_recognition.drawing_app import run_drawing_app
from number_recognition.model import load_model

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src" / "number_recognition").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
MODEL_PATH = PROJECT_ROOT / "models" / "digit_cnn.pth"

print(f"PyTorch : {torch.__version__}")
print(f"Appareil : {DEVICE}")
print(f"Modèle : {MODEL_PATH}")
```

## 1. Charger le CNN entraîné

L'architecture `DigitCNN` et le chargement du checkpoint sont définis dans `src/number_recognition/model.py` et réutilisés ici.

```python
if not MODEL_PATH.is_file():
    raise FileNotFoundError(
        f"Modèle introuvable : {MODEL_PATH}. "
        "Exécute le notebook d'entraînement et sa cellule de sauvegarde."
    )

model = load_model(MODEL_PATH, DEVICE)
print("Modèle chargé.")
```

## 2. Dessiner un chiffre

La fenêtre fournit un canevas blanc avec un pinceau noir. L'image dessinée est automatiquement inversée, convertie en niveaux de gris et redimensionnée en `28 × 28`, comme les images MNIST. Clique sur **Effacer** pour recommencer.

```python
run_drawing_app(model, DEVICE)
```

## 3. Prédire le chiffre
